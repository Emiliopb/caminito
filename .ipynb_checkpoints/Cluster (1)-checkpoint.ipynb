```python
def latitude(texto):
    from geopy.geocoders import Nominatim
    loc = Nominatim(user_agent="GetLoc")
    try: 
        # entering the location name
        getLoc = loc.geocode(texto)
    except:
        
    return getLoc.latitude
```

```python
df_datos['Latitud'] = df_datos['Origen'].apply(latitude)
```
